```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def image_enhancement(image):
  # Convert the image to grayscale (optional, if it's a colored image)
  gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

  # Apply Non-Local Means Denoising
  denoised_image = cv2.fastNlMeansDenoisingColored(image, None, 10, 10, 7, 21)

  # Optional: Apply a Gaussian Blur to further smooth the image
  blurred_image = cv2.GaussianBlur(denoised_image, (3, 3), 0)

  # Sharpen the image using a kernel
  # Define a kernel for sharpening
  sharpening_kernel = np.array([[-1, -1, -1],
                                [-1,  9, -1],
                                [-1, -1, -1]])

  sharpened_image = cv2.filter2D(blurred_image, -1, sharpening_kernel)

  return sharpened_image
```

```python
# Image 1
image1 = cv2.imread('grainy_girls.jpg', cv2.IMREAD_COLOR)
sharpened_image1 = image_enhancement(image1)

# Display the original and enhanced images
fig, ax = plt.subplots(1, 2, figsize=(12, 6))

ax[0].imshow(cv2.cvtColor(image1, cv2.COLOR_BGR2RGB))
ax[0].set_title("Original Grainy Image")
ax[0].axis('off')

ax[1].imshow(cv2.cvtColor(sharpened_image1, cv2.COLOR_BGR2RGB))
ax[1].set_title("Enhanced Image")
ax[1].axis('off')
```

```python
# Image 2
image2 = cv2.imread('grainy_img.jpg', cv2.IMREAD_COLOR)
sharpened_image2 = image_enhancement(image2)

# Display the original and enhanced images
fig, ax = plt.subplots(1, 2, figsize=(12, 6))

ax[0].imshow(cv2.cvtColor(image2, cv2.COLOR_BGR2RGB))
ax[0].set_title("Original Grainy Image")
ax[0].axis('off')

ax[1].imshow(cv2.cvtColor(sharpened_image2, cv2.COLOR_BGR2RGB))
ax[1].set_title("Enhanced Image")
ax[1].axis('off')
```
